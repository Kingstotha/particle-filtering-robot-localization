```python
pacman_errors = [3.03, 4.21, 2.27, 2.10, 2.81,
                 2.26, 2.32, 3.63, 2.53, 2.44]

ghost_errors = [7.40, 6.69, 34.91, 5.99, 37.61,
                5.64, 7.43, 8.50, 9.02, 6.78]


pacman_mean_errors = sum(pacman_errors)/len(pacman_errors)

ghost_mean_erros = sum(ghost_errors)/len(ghost_errors)

pacman_squared_deviations = [(i - pacman_mean_errors)**2 for i in pacman_errors]
ghost_squared_deviations = [(i - ghost_mean_erros)**2 for i in ghost_errors]

pacman_variance = sum(pacman_squared_deviations)/len(pacman_squared_deviations)
ghost_variance = sum(ghost_squared_deviations)/len(ghost_squared_deviations)

pacman_std = pacman_variance**0.5
ghost_std = ghost_variance**0.5

pacman_min = min(pacman_errors)
pacman_max = max(pacman_errors)
ghost_min = min(ghost_errors)
ghost_max = max(ghost_errors)

print("Pacman mean error:", pacman_mean_errors)
print("Ghost mean error:", ghost_mean_erros)
print()
print("Pacman Squared devistions:", pacman_squared_deviations)
print("Pacman Variance:", pacman_variance)
print("Pacman Standard deviation:", pacman_std)
print()
print("Ghost squared deviations:", ghost_squared_deviations)
print("Ghost Variance:", ghost_variance)
print("Ghost Standard Deviation:", ghost_std)
print()
print("Pacman minimum error:", pacman_min)
print("Pacman maximum error:", pacman_max)
print("Ghost minimum error:", ghost_min)
print("Ghost maximum error:", ghost_max)
```
